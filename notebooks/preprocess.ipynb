### 2.3 Kiểm tra ngoại lai bằng phương pháp IQR

```python
# Chọn các cột số để kiểm tra ngoại lai
numeric_cols = df_clean.select_dtypes(include=["int64", "float64"]).columns

outlier_summary = []

for col in numeric_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outlier_count = ((df_clean[col] < lower_bound) | (df_clean[col] > upper_bound)).sum()
    outlier_percent = outlier_count / len(df_clean) * 100
    
    outlier_summary.append({
        "column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound,
        "outlier_count": outlier_count,
        "outlier_percent": outlier_percent,
        "outlier_percent_text": f"{outlier_percent:.2f}%"
    })

outlier_summary_df = pd.DataFrame(outlier_summary)
outlier_summary_df = outlier_summary_df.sort_values(by="outlier_count", ascending=False)

print("Bảng tổng hợp ngoại lai theo phương pháp IQR:")
display(outlier_summary_df)
```

Kết quả kiểm tra ngoại lai bằng phương pháp IQR cho thấy một số biến có tỷ lệ ngoại lai khá cao. Trong đó, `mazut_bog_spending` có tỷ lệ ngoại lai cao nhất với 512 dòng, chiếm 17.38%; tiếp theo là `diesel_bog_spending` với 403 dòng, chiếm 13.68%. Một số biến giá xăng dầu như `e5ron92_retail_price`, `ron95_base_price`, `mazut_retail_price` cũng có tỷ lệ ngoại lai đáng chú ý, dao động từ khoảng 10% đến 13%.

Các ngoại lai này không được loại bỏ khỏi dữ liệu vì chúng có thể phản ánh biến động thực tế của thị trường xăng dầu, giá dầu thế giới, tỷ giá hoặc chính sách điều hành giá trong từng giai đoạn. Đặc biệt, các biến liên quan đến Quỹ bình ổn giá thường có nhiều giá trị bằng 0 và chỉ phát sinh mức trích lập/chi quỹ cao ở một số thời điểm, nên phương pháp IQR dễ đánh dấu các giai đoạn này là ngoại lai.

Biến `is_adjustment_day` cũng được phát hiện có 247 ngoại lai, chiếm 8.38%. Tuy nhiên, đây là biến nhị phân dùng để xác định ngày có điều chỉnh giá hay không, không phải biến số liên tục, nên không xem đây là ngoại lai cần xử lý.

Vì vậy, nhóm quyết định giữ nguyên các giá trị ngoại lai thay vì xóa bỏ. Việc giữ lại các giá trị này giúp bảo toàn các giai đoạn biến động quan trọng, phục vụ cho việc phân tích xu hướng, mức độ biến động và tác động của các yếu tố thị trường/chính sách đến giá xăng dầu tại Việt Nam.


### 2.4 Chuẩn hóa kiểu dữ liệu

```python
# Chuyển cột date sang kiểu datetime
df_clean["date"] = pd.to_datetime(df_clean["date"], dayfirst=True, errors="coerce")

# Kiểm tra lại số dòng bị lỗi ngày tháng sau khi chuyển đổi
invalid_date_count = df_clean["date"].isna().sum()
print("Số dòng có ngày tháng không hợp lệ:", invalid_date_count)

# Chuyển is_adjustment_day sang kiểu boolean
df_clean["is_adjustment_day"] = df_clean["is_adjustment_day"].astype(bool)

# Chuyển các cột còn lại sang dạng số nếu cần
numeric_cols = df_clean.columns.drop(["date", "is_adjustment_day"])

for col in numeric_cols:
    df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce")

# Sắp xếp dữ liệu theo thời gian
df_clean = df_clean.sort_values("date").reset_index(drop=True)

# Kiểm tra lại kiểu dữ liệu sau khi chuẩn hóa
print("Kiểu dữ liệu sau khi chuẩn hóa:")
display(df_clean.dtypes)
```

Sau khi chuẩn hóa, cột date đã được chuyển sang kiểu `datetime64[ns]` và không có dòng nào bị lỗi ngày tháng, với số dòng có ngày tháng không hợp lệ bằng 0. Điều này cho thấy dữ liệu thời gian hợp lệ và có thể sử dụng để phân tích xu hướng theo ngày, tháng, quý hoặc năm.

Bên cạnh đó biến `is_adjustment_day` được chuyển sang kiểu bool, phù hợp với ý nghĩa của biến là xác định ngày có điều chỉnh giá hay không.

### 2.5 Kiểm tra logic ở biến `is_adjustment_day`
Biến `is_adjustment`_day là biến nhị phân dùng để xác định một ngày có phải là ngày điều chỉnh giá xăng dầu hay không. Nên cần kiểm tra logic của biến này vì bộ dữ liệu được ghi nhận theo ngày, trong khi giá xăng dầu không thay đổi hằng ngày mà chỉ thay đổi vào các kỳ điều chỉnh. Nếu biến `is_adjustment_day` bị gán sai, các phân tích về số lần tăng/giảm giá, mức điều chỉnh giá và các giai đoạn biến động sẽ không chính xác.

```python
# Các cột liên quan trực tiếp đến điều chỉnh giá trong nước
price_policy_cols = [
    "diesel_base_price",
    "diesel_retail_price",
    "diesel_bog_contribution",
    "diesel_bog_spending",
    "e5ron92_base_price",
    "e5ron92_retail_price",
    "e5ron92_bog_contribution",
    "e5ron92_bog_spending",
    "mazut_base_price",
    "mazut_retail_price",
    "mazut_bog_contribution",
    "mazut_bog_spending",
    "ron95_base_price",
    "ron95_retail_price",
    "ron95_bog_contribution",
    "ron95_bog_spending"
]

# Sắp xếp theo ngày
df_check = df_clean.sort_values("date").reset_index(drop=True)

# Kiểm tra xem ngày hiện tại có thay đổi giá/chính sách so với ngày trước đó không
df_check["has_price_policy_change"] = (
    df_check[price_policy_cols]
    .diff()
    .abs()
    .sum(axis=1) > 0
)

# Dòng đầu tiên không có ngày trước để so sánh, có thể xem là mốc bắt đầu
df_check.loc[0, "has_price_policy_change"] = df_check.loc[0, "is_adjustment_day"]

# So sánh biến gốc với biến tự tính
df_check["is_match"] = df_check["is_adjustment_day"] == df_check["has_price_policy_change"]

# Tổng kết kết quả kiểm tra
print("Tổng số dòng:", len(df_check))
print("Số dòng khớp:", df_check["is_match"].sum())
print("Số dòng không khớp:", (~df_check["is_match"]).sum())
print("Tỷ lệ khớp:", round(df_check["is_match"].mean() * 100, 2), "%")
```

Kiểm tra tính hợp lệ của biến `is_adjustment_day` bằng cách so sánh với sự thay đổi của các biến giá bán lẻ, giá cơ sở và các biến liên quan đến Quỹ bình ổn giá so với ngày liền trước. Kết quả cho thấy toàn bộ 2.946 dòng đều khớp, không có dòng nào không khớp, đạt tỷ lệ khớp 100%.

```python
# Kiểm chứng cách hiểu về mối quan hệ giữa 4 biến: base_price, bog_contribution, bog_spending, retail_price
products = ["diesel", "e5ron92", "mazut", "ron95"]

check_relation = pd.DataFrame({
    "product": products,
    "so_dong_khong_khop(base + contribution - spending = retail)": [
        (
            (df_clean[f"{p}_base_price"]
             + df_clean[f"{p}_bog_contribution"]
             - df_clean[f"{p}_bog_spending"]
             - df_clean[f"{p}_retail_price"]).abs() > 0.01
        ).sum()
        for p in products
    ]
})
display(check_relation)
```

Kết quả đối chiếu công thức `giá bán lẻ = giá cơ sở + mức trích lập Quỹ BOG − mức chi Quỹ BOG` cho từng mặt hàng cho thấy cả 4 mặt hàng đều khớp trên toàn bộ 2.946 dòng, qua đó xác nhận đúng mối quan hệ giữa 4 biến giá trong cùng một nhóm mặt hàng. Về ý nghĩa nghiệp vụ, công thức này cũng hợp lý: khi Nhà nước trích lập vào Quỹ BOG, phần trích lập đó được cộng thêm vào giá bán lẻ (người mua trả thêm để góp vào quỹ); ngược lại khi Nhà nước chi Quỹ BOG để bình ổn giá, phần chi đó được trừ bớt khỏi giá bán lẻ (người mua được hỗ trợ giảm giá từ quỹ).

```python
# Thống kê nhanh số kỳ điều chỉnh giá và khoảng cách trung bình giữa các kỳ điều chỉnh
adjustment_dates = df_clean.loc[df_clean["is_adjustment_day"], "date"].sort_values()
gap_days = adjustment_dates.diff().dt.days.dropna()

print("Tổng số kỳ điều chỉnh giá (is_adjustment_day = True):", df_clean["is_adjustment_day"].sum())
print("Khoảng cách trung bình giữa 2 kỳ điều chỉnh liên tiếp:", round(gap_days.mean(), 1), "ngày")
print("Khoảng cách nhỏ nhất / lớn nhất giữa 2 kỳ điều chỉnh:", gap_days.min(), "/", gap_days.max(), "ngày")
```

### 4.3 Kiểm tra tính liên tục và tần suất của trục thời gian

```python
# Khoảng thời gian của dữ liệu
date_min = df_clean["date"].min()
date_max = df_clean["date"].max()
n_days_expected = (date_max - date_min).days + 1

print("Ngày bắt đầu:", date_min.date())
print("Ngày kết thúc:", date_max.date())
print("Số ngày kỳ vọng nếu liên tục (không thiếu ngày nào):", n_days_expected)
print("Số dòng thực tế trong dữ liệu:", df_clean.shape[0])
print("Số ngày bị thiếu trong chuỗi thời gian:", n_days_expected - df_clean.shape[0])

if n_days_expected - df_clean.shape[0] > 0:
# Kiểm tra biến date có bị trùng lặp không (mỗi ngày chỉ nên xuất hiện 1 lần)
    print("\nSố ngày bị trùng lặp (xuất hiện nhiều hơn 1 dòng):", df_clean["date"].duplicated().sum())

    # Tìm cụ thể các ngày bị thiếu (nếu có)
    full_range = pd.date_range(start=date_min, end=date_max, freq="D")
    missing_dates = full_range.difference(df_clean["date"])
    print("Số lượng ngày bị thiếu cụ thể:", len(missing_dates))
    if len(missing_dates) > 0:
        print("Một số ngày bị thiếu:", list(missing_dates[:10].date))
```
